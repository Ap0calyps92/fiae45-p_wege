```sql
SELECT
    p.PatientNr,
    p.Nachname,
    p.Vorname,
    z.Bezeichnung AS Zusatzleistung,
    ez.LeistungVon,
    ez.LeistungBis
FROM ErbrachteZusatzleistung ez
JOIN Patient p ON ez.PatientNr = p.PatientNr
JOIN ZusatzLeistung z ON ez.ZusatzLeistungNr = z.ZusatzLeistungNr;
```

```sql
CREATE VIEW vw_ErbrachteZusatzleistungen AS
SELECT
    p.PatientNr,
    p.Nachname,
    p.Vorname,
    z.Bezeichnung AS Zusatzleistung,
    ez.LeistungVon,
    ez.LeistungBis
FROM ErbrachteZusatzleistung ez
JOIN Patient p ON ez.PatientNr = p.PatientNr
JOIN ZusatzLeistung z ON ez.ZusatzLeistungNr = z.ZusatzLeistungNr;
```

```sql
-- Test ob View geklappt hat
SELECT * FROM vw_ErbrachteZusatzleistungen;
```

```sql
SELECT
    Zusatzleistung,
    LeistungVon,
    LeistungBis
FROM vw_ErbrachteZusatzleistungen
WHERE Vorname = 'Klaus' AND Nachname = 'Müller'
ORDER BY LeistungVon, Zusatzleistung;
```

```sql
SELECT
    p.PatientNr,
    p.Nachname,
    p.Vorname,
    eml.Leistungsdatum,
    ml.Bezeichnung AS MedizinischeLeistung,
    a.Name AS Arzt
FROM ErbrachteMedLeistung eml
JOIN Patient p ON eml.PatientNr = p.PatientNr
JOIN MedLeistung ml ON eml.MedLeistungNr = ml.MedLeistungNr
JOIN Arzt a ON eml.ArztNr = a.ArztNr;
```

```sql
CREATE VIEW vw_ErteilteMedileistung AS
SELECT
    p.PatientNr,
    p.Nachname,
    p.Vorname,
    eml.Leistungsdatum,
    ml.Bezeichnung AS MedizinischeLeistung,
    a.Name AS Arzt
FROM ErbrachteMedLeistung eml
JOIN Patient p ON eml.PatientNr = p.PatientNr
JOIN MedLeistung ml ON eml.MedLeistungNr = ml.MedLeistungNr
JOIN Arzt a ON eml.ArztNr = a.ArztNr;
```

```sql
SELECT
    Leistungsdatum,
    MedizinischeLeistung,
    Arzt
FROM vw_ErteilteMedileistung
WHERE Vorname = 'Britta' AND Nachname = 'Schulz'
ORDER BY Leistungsdatum;
```
