```sql
SELECT
    ROW_NUMBER() OVER (ORDER BY f.Fahrt_ID) AS Fahrt,
    f.Nutzer_ID AS Nutzer,
    ft.Bezeichnung AS Fahrzeug_Typ,
    fz.Kennzeichen,
    CONCAT(FORMAT(f.Abholung_Datum, 'dd.MM.'), '/', FORMAT(f.Abholung_Zeit, 'HH:mm')) AS Abholung_Datum_Zeit,
    CONCAT(FORMAT(f.Rueckgabe_Datum, 'dd.MM.'), '/', FORMAT(f.Rueckgabe_Zeit, 'HH:mm')) AS Rueckgabe_Datum_Zeit,
    f.Dauer_Std AS Dauer_in_Std,
    ft.EUR_Std,
    CAST(f.Dauer_Std * ft.EUR_Std AS NUMERIC(9,2)) AS Std_in_EUR,
    FORMAT(f.Start_km, 'N0', 'de-DE') AS Start_km,
    FORMAT(f.Ende_km, 'N0', 'de-DE') AS Ende_km,
    f.Strecke_km AS Strecke_in_km,
    ft.EUR_km,
    CAST(f.Strecke_km * ft.EUR_km AS NUMERIC(9,2)) AS km_in_EUR
FROM FAHRT f
JOIN NUTZER n ON f.Nutzer_ID = n.Nutzer_ID
JOIN FAHRZEUG fz ON f.Fahrzeug_ID = fz.Fahrzeug_ID
JOIN FAHRZEUGTYP ft ON fz.Fahrzeugtyp_ID = ft.Fahrzeugtyp_ID
WHERE f.Rechnungs_Nr = '102015/005498'
ORDER BY f.Fahrt_ID;
```

```sql
SELECT
    ROW_NUMBER() OVER (ORDER BY f.Fahrt_ID) AS Fahrt,
    f.Nutzer_ID AS Nutzer,
    ft.Bezeichnung AS Fahrzeug_Typ,
    fz.Kennzeichen,
    CONCAT(FORMAT(f.Abholung, 'dd.MM.'), '/', FORMAT(f.Abholung, 'HH:mm')) AS Abholung_Datum_Zeit,
    CONCAT(FORMAT(f.Rueckgabe, 'dd.MM.'), '/', FORMAT(f.Rueckgabe, 'HH:mm')) AS Rueckgabe_Datum_Zeit,
    f.Dauer_Std AS Dauer_in_Std,
    f.EUR_Std,
    CAST(f.Dauer_Std * f.EUR_Std AS NUMERIC(9,2)) AS Std_in_EUR,
    FORMAT(f.Start_km, 'N0', 'de-DE') AS Start_km,
    FORMAT(f.Ende_km, 'N0', 'de-DE') AS Ende_km,
    f.Strecke_km AS Strecke_in_km,
    f.EUR_km,
    CAST(f.Strecke_km * f.EUR_km AS NUMERIC(9,2)) AS km_in_EUR
FROM FAHRT f
JOIN NUTZER n ON f.Nutzer_ID = n.Nutzer_ID
JOIN FAHRZEUG fz ON f.Fahrzeug_ID = fz.Fahrzeug_ID
JOIN FAHRZEUGTYP ft ON fz.Fahrzeugtyp_ID = ft.Fahrzeugtyp_ID
WHERE n.Kunden_Nr = 'K10200'
  AND f.Abholung BETWEEN '2015-10-01' AND '2015-10-31 23:59:59'
ORDER BY f.Fahrt_ID;
```
