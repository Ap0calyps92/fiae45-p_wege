# Datenbank-Objekte

Dieses Notebook verwendet SQL.



```sql
USE master;

DROP DATABASE IF EXISTS Carsharing;
GO

CREATE DATABASE Carsharing;
GO

USE Carsharing;

CREATE TABLE KUNDE
(
  Kunden_Nr VARCHAR(100) NOT NULL,
  Name VARCHAR(100) NOT NULL,
  Strasse VARCHAR(100) NOT NULL,
  PLZ CHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_KUNDE PRIMARY KEY (Kunden_Nr)
);

CREATE TABLE NUTZER
(
  Nutzer_ID VARCHAR(100) NOT NULL,
  Kunden_Nr VARCHAR(100) NOT NULL,
  CONSTRAINT pk_NUTZER PRIMARY KEY (Nutzer_ID),
  CONSTRAINT fk_NUTZER_KUNDE FOREIGN KEY (Kunden_Nr) REFERENCES KUNDE(Kunden_Nr)
);

CREATE TABLE RECHNUNG
(
  Rechnungs_Nr VARCHAR(100) NOT NULL,
  Rechnungsdatum DATE NOT NULL,
  Zeitraum_von DATE NOT NULL,
  Zeitraum_bis DATE NOT NULL,
  Monatspauschale NUMERIC(9,2) NOT NULL,
  Kunden_Nr VARCHAR(100) NOT NULL,
  CONSTRAINT pk_RECHNUNG PRIMARY KEY (Rechnungs_Nr),
  CONSTRAINT fk_RECHNUNG_KUNDE FOREIGN KEY (Kunden_Nr) REFERENCES KUNDE(Kunden_Nr)
);

CREATE TABLE FAHRZEUGTYP
(
  Fahrzeugtyp_ID INT IDENTITY(1,1) NOT NULL,
  Bezeichnung VARCHAR(100) NOT NULL,
  EUR_Std NUMERIC(9,2) NOT NULL,
  EUR_km NUMERIC(9,2) NOT NULL,
  CONSTRAINT pk_FAHRZEUGTYP PRIMARY KEY (Fahrzeugtyp_ID)
);

CREATE TABLE FAHRZEUG
(
  Fahrzeug_ID INT IDENTITY(1,1) NOT NULL,
  Kennzeichen VARCHAR(100) NOT NULL,
  Fahrzeugtyp_ID INT NOT NULL,
  CONSTRAINT pk_FAHRZEUG PRIMARY KEY (Fahrzeug_ID),
  CONSTRAINT fk_FAHRZEUG_FAHRZEUGTYP FOREIGN KEY (Fahrzeugtyp_ID) REFERENCES FAHRZEUGTYP(Fahrzeugtyp_ID)
);

CREATE TABLE FAHRT
(
  Fahrt_ID INT IDENTITY(1,1) NOT NULL,
  Abholung_Datum DATE NOT NULL,
  Abholung_Zeit TIME NOT NULL,
  Rueckgabe_Datum DATE NOT NULL,
  Rueckgabe_Zeit TIME NOT NULL,
  Dauer_Std FLOAT NOT NULL,
  Start_km INT NOT NULL,
  Ende_km INT NOT NULL,
  Strecke_km INT NOT NULL,
  Rechnungs_Nr VARCHAR(100) NOT NULL,
  Nutzer_ID VARCHAR(100) NOT NULL,
  Fahrzeug_ID INT NOT NULL,
  CONSTRAINT pk_FAHRT PRIMARY KEY (Fahrt_ID),
  CONSTRAINT fk_FAHRT_RECHNUNG FOREIGN KEY (Rechnungs_Nr) REFERENCES RECHNUNG(Rechnungs_Nr),
  CONSTRAINT fk_FAHRT_NUTZER FOREIGN KEY (Nutzer_ID) REFERENCES NUTZER(Nutzer_ID),
  CONSTRAINT fk_FAHRT_FAHRZEUG FOREIGN KEY (Fahrzeug_ID) REFERENCES FAHRZEUG(Fahrzeug_ID)
);
```
