Dieses Notebook verwendet SQL.



```sql
USE master;
GO

IF DB_ID('MaeckFitt') IS NOT NULL
BEGIN
    ALTER DATABASE MaeckFitt SET SINGLE_USER WITH ROLLBACK IMMEDIATE;
    DROP DATABASE MaeckFitt;
END
GO

CREATE DATABASE MaeckFitt;
GO

USE MaeckFitt;
GO

-- ============================================
-- Tabellen löschen (in Reihenfolge der Abhängigkeiten,
-- "Kind"-Tabellen zuerst wegen FOREIGN KEYs)
-- ============================================

IF OBJECT_ID('dbo.KURSART_TRAININGSTAG', 'U') IS NOT NULL DROP TABLE dbo.KURSART_TRAININGSTAG;
IF OBJECT_ID('dbo.BUCHT', 'U') IS NOT NULL DROP TABLE dbo.BUCHT;
IF OBJECT_ID('dbo.TRAININGSTAG', 'U') IS NOT NULL DROP TABLE dbo.TRAININGSTAG;
IF OBJECT_ID('dbo.TRAININGSSTAND', 'U') IS NOT NULL DROP TABLE dbo.TRAININGSSTAND;
IF OBJECT_ID('dbo.KURS', 'U') IS NOT NULL DROP TABLE dbo.KURS;
IF OBJECT_ID('dbo.KUNDE', 'U') IS NOT NULL DROP TABLE dbo.KUNDE;
IF OBJECT_ID('dbo.KURSLEITER', 'U') IS NOT NULL DROP TABLE dbo.KURSLEITER;
IF OBJECT_ID('dbo.KURSART', 'U') IS NOT NULL DROP TABLE dbo.KURSART;
IF OBJECT_ID('dbo.TRAININGSART', 'U') IS NOT NULL DROP TABLE dbo.TRAININGSART;
GO

-- ============================================
-- Basistabellen (ohne Abhängigkeiten)
-- ============================================

CREATE TABLE TRAININGSART
(
    Trainingsart_ID INT NOT NULL,
    Name            VARCHAR(100) NOT NULL,
    CONSTRAINT PK_Trainingsart PRIMARY KEY (Trainingsart_ID)
);
GO

CREATE TABLE KURSLEITER
(
    Kursleiter_ID INT NOT NULL,
    Name          VARCHAR(100) NOT NULL,
    CONSTRAINT PK_Kursleiter PRIMARY KEY (Kursleiter_ID)
);
GO

CREATE TABLE KUNDE
(
    Kunde_ID INT NOT NULL,
    Name     VARCHAR(100) NOT NULL,
    Passwort VARCHAR(255) NOT NULL,
    CONSTRAINT PK_Kunde PRIMARY KEY (Kunde_ID)
);
GO

CREATE TABLE TRAININGSTAG
(
    Name   VARCHAR(100) NOT NULL,
    Kürzel CHAR(2) NOT NULL,
    CONSTRAINT PK_Trainingstag PRIMARY KEY (Kürzel)
);
GO

-- ============================================
-- Tabellen mit Abhängigkeiten
-- ============================================

CREATE TABLE KURSART
(
    Kursart_ID        INT NOT NULL,
    Beschreibung      VARCHAR(100) NOT NULL,
    Trainingsstunden  INT NOT NULL,
    max_Teilnehmer    INT NOT NULL,
    Preis             NUMERIC(9,2) NOT NULL,
    Kürzel            CHAR(2) NOT NULL,
    Trainingsart_ID   INT NOT NULL,
    CONSTRAINT PK_Kursart PRIMARY KEY (Kursart_ID),
    CONSTRAINT FK_Kursart_Trainingsart FOREIGN KEY (Trainingsart_ID)
        REFERENCES TRAININGSART(Trainingsart_ID)
);
GO

CREATE TABLE KURS
(
    Kurs_ID       INT NOT NULL,
    Startdatum    DATE NOT NULL,
    Zeit_von      TIME NOT NULL,
    Zeit_bis      TIME NOT NULL,
    Status        BIT NOT NULL,
    Kursart_ID    INT NOT NULL,
    Kursleiter_ID INT NOT NULL,
    CONSTRAINT PK_Kurs PRIMARY KEY (Kurs_ID),
    CONSTRAINT FK_Kurs_Kursart FOREIGN KEY (Kursart_ID)
        REFERENCES KURSART(Kursart_ID),
    CONSTRAINT FK_Kurs_Kursleiter FOREIGN KEY (Kursleiter_ID)
        REFERENCES KURSLEITER(Kursleiter_ID)
);
GO

CREATE TABLE TRAININGSSTAND
(
    Datum           DATE NOT NULL,
    Körpergewicht   NUMERIC(5,2) NOT NULL,
    Fitnessgrad     NUMERIC(5,2) NOT NULL,
    Kunde_ID        INT NOT NULL,
    CONSTRAINT PK_Trainingsstand PRIMARY KEY (Datum, Kunde_ID),
    CONSTRAINT FK_Trainingsstand_Kunde FOREIGN KEY (Kunde_ID)
        REFERENCES KUNDE(Kunde_ID)
);
GO

CREATE TABLE BUCHT
(
    Buchungsdatum DATE NULL,
    Kunde_ID      INT NOT NULL,
    Kurs_ID       INT NOT NULL,
    CONSTRAINT PK_Bucht PRIMARY KEY (Kunde_ID, Kurs_ID),
    CONSTRAINT FK_Bucht_Kunde FOREIGN KEY (Kunde_ID)
        REFERENCES KUNDE(Kunde_ID),
    CONSTRAINT FK_Bucht_Kurs FOREIGN KEY (Kurs_ID)
        REFERENCES KURS(Kurs_ID)
);
GO

CREATE TABLE KURSART_TRAININGSTAG
(
    Kürzel     CHAR(2) NOT NULL,
    Kursart_ID INT NOT NULL,
    CONSTRAINT PK_Kursart_Trainingstag PRIMARY KEY (Kürzel, Kursart_ID),
    CONSTRAINT FK_KursartTrainingstag_Trainingstag FOREIGN KEY (Kürzel)
        REFERENCES TRAININGSTAG(Kürzel),
    CONSTRAINT FK_KursartTrainingstag_Kursart FOREIGN KEY (Kursart_ID)
        REFERENCES KURSART(Kursart_ID)
);
GO

-- ============================================
-- Kontrolle: alle Tabellen anzeigen
-- ============================================

SELECT TABLE_NAME
FROM INFORMATION_SCHEMA.TABLES
WHERE TABLE_TYPE = 'BASE TABLE'
ORDER BY TABLE_NAME;
GO
```
