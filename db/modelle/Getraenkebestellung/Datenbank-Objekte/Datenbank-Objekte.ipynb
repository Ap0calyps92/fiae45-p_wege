# Datenbank-Objekte

Dieses Notebook verwendet SQL.



```sql
USE master;

DROP DATABASE IF EXISTS Getraenkebestellung;
GO

CREATE DATABASE Getraenkebestellung;
GO

USE Getraenkebestellung;

CREATE TABLE LIEFERANT
(
  Lieferant_ID INT IDENTITY(1,1) NOT NULL,
  Name VARCHAR(100) NOT NULL,
  Straße VARCHAR(100) NOT NULL,
  PLZ VARCHAR(100) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_LIEFERANT PRIMARY KEY (Lieferant_ID)
);

CREATE TABLE BESTELLUNG
(
  Bestell_ID INT IDENTITY(1,1) NOT NULL,
  Datum DATE NOT NULL,
  Lieferant_ID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG PRIMARY KEY (Bestell_ID),
  CONSTRAINT fk_BESTELLUNG_LIEFERANT FOREIGN KEY (Lieferant_ID) REFERENCES LIEFERANT(Lieferant_ID)
);

CREATE TABLE ARTIKEL
(
  Artikel_ID INT IDENTITY(1,1) NOT NULL,
  Bezeichnung VARCHAR(100) NOT NULL,
  UVP NUMERIC(9,2) NOT NULL,
  Gebinde VARCHAR(100) NOT NULL,
  CONSTRAINT pk_ARTIKEL PRIMARY KEY (Artikel_ID)
);

CREATE TABLE ENTHAELT
(
  Menge INT NOT NULL,
  Einzelpreis NUMERIC(9,2) NOT NULL,
  Bestell_ID INT NOT NULL,
  Artikel_ID INT NOT NULL,
  CONSTRAINT pk_ENTHAELT PRIMARY KEY (Bestell_ID, Artikel_ID),
  CONSTRAINT fk_ENTHAELT_BESTELLUNG FOREIGN KEY (Bestell_ID) REFERENCES BESTELLUNG(Bestell_ID),
  CONSTRAINT fk_ENTHAELT_ARTIKEL FOREIGN KEY (Artikel_ID) REFERENCES ARTIKEL(Artikel_ID)
);

CREATE TABLE BIETET_AN
(
  Angebotspreis NUMERIC(9,2) NOT NULL,
  Lieferant_ID INT NOT NULL,
  Artikel_ID INT NOT NULL,
  CONSTRAINT pk_BIETET_AN PRIMARY KEY (Lieferant_ID, Artikel_ID),
  CONSTRAINT fk_BIETET_AN_LIEFERANT FOREIGN KEY (Lieferant_ID) REFERENCES LIEFERANT(Lieferant_ID),
  CONSTRAINT fk_BIETET_AN_ARTIKEL FOREIGN KEY (Artikel_ID) REFERENCES ARTIKEL(Artikel_ID)
);
```
