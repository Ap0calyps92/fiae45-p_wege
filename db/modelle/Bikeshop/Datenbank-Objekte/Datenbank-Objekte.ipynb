# Datenbank-Objekte

Der von ERDPlus generierte Code wird erweitert.

- Erzeugen der Datenbank.
- Evlt. Löschen der alten Version.
- Umschalten der Datenbank, in die geschrieben wird.
- Für PKs und FKs ```CONSTRAINT``` Syntax.

````sql
USE master;

DROP DATABASE IF EXISTS Bikeshop_DEMO;
-- Batch Operator, teilt Befehle in Portionen auf
GO

CREATE DATABASE Bikeshop_DEMO;
GO

USE Bikeshop_DEMO;

CREATE TABLE KUNDE
(
  KundenNr INT NOT NULL,
  Firma VARCHAR(100),
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  Straße VARCHAR(100) NOT NULL,
  PLZ VARCHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_KUNDE PRIMARY KEY (KundenNr)
);USE master;

DROP DATABASE IF EXISTS Bikeshop_DEMO;
-- Batch Operator, teilt Befehle in Portionen auf
GO

CREATE DATABASE Bikeshop_DEMO;
GO

USE Bikeshop_DEMO;

CREATE TABLE KUNDE
(
  KundenNr INT NOT NULL,
  Firma VARCHAR(100),
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  Straße VARCHAR(100) NOT NULL,
  PLZ VARCHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_KUNDE PRIMARY KEY (KundenNr)
);

CREATE TABLE MITARBEITER
(
  MitarbeiterID INT NOT NULL,
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  CONSTRAINT pk_MITARBEITER PRIMARY KEY (MitarbeiterID)
);

CREATE TABLE BESTELLUNG
(
  BestellungID INT NOT NULL,
  Bestelldatum DATE NOT NULL,
  Lieferdatum DATE,
  KundenNr INT NOT NULL,
  MitarbeiterID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG PRIMARY KEY (BestellungID),
  CONSTRAINT fk_BESTELLUNG_KUNDE FOREIGN KEY (KundenNr) REFERENCES KUNDE(KundenNr),
  CONSTRAINT fk_BESTELLUNG_MITARBEITER FOREIGN KEY (MitarbeiterID) REFERENCES MITARBEITER(MitarbeiterID)
);

CREATE TABLE PRODUKTKATEGORIE
(
  ProduktkategorieID INT NOT NULL,
  Kategorie VARCHAR(100) NOT NULL,
  CONSTRAINT pk_PRODUKTKATEGORIE PRIMARY KEY (ProduktkategorieID)
);

CREATE TABLE PRODUKT
(
  ProduktID INT NOT NULL,
  Produktname VARCHAR(100) NOT NULL,
  Verkaufspreis NUMERIC(9,2) NOT NULL,
  Einkaufspreis NUMERIC(9,2) NOT NULL,
  ProduktkategorieID INT NOT NULL,
  CONSTRAINT pk_PRODUKT PRIMARY KEY (ProduktID),
  CONSTRAINT fk_PRODUKT_PRODUKTKATEGORIE FOREIGN KEY (ProduktkategorieID) REFERENCES PRODUKTKATEGORIE(ProduktkategorieID)
);

CREATE TABLE BESTELLUNG_PRODUKT
(
  Menge INT NOT NULL,
  Einzelpreis NUMERIC(9,2) NOT NULL,
  BestellungID INT NOT NULL,
  ProduktID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG_PRODUKT PRIMARY KEY (BestellungID, ProduktID),
  CONSTRAINT fk_BESTELLUNG_PRODUKT_BESTELLUNG FOREIGN KEY (BestellungID) REFERENCES BESTELLUNG(BestellungID),
  CONSTRAINT fk_BESTELLUNG_PRODUKT_PRODUKT FOREIGN KEY (ProduktID) REFERENCES PRODUKT(ProduktID)
);USE master;

DROP DATABASE IF EXISTS Bikeshop_DEMO;
-- Batch Operator, teilt Befehle in Portionen auf
GO

CREATE DATABASE Bikeshop_DEMO;
GO

USE Bikeshop_DEMO;

CREATE TABLE KUNDE
(
  KundenNr INT NOT NULL,
  Firma VARCHAR(100),
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  Straße VARCHAR(100) NOT NULL,
  PLZ VARCHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_KUNDE PRIMARY KEY (KundenNr)
);

CREATE TABLE MITARBEITER
(
  MitarbeiterID INT NOT NULL,
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  CONSTRAINT pk_MITARBEITER PRIMARY KEY (MitarbeiterID)
);

CREATE TABLE BESTELLUNG
(
  BestellungID INT NOT NULL,
  Bestelldatum DATE NOT NULL,
  Lieferdatum DATE,
  KundenNr INT NOT NULL,
  MitarbeiterID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG PRIMARY KEY (BestellungID),
  CONSTRAINT fk_BESTELLUNG_KUNDE FOREIGN KEY (KundenNr) REFERENCES KUNDE(KundenNr),
  CONSTRAINT fk_BESTELLUNG_MITARBEITER FOREIGN KEY (MitarbeiterID) REFERENCES MITARBEITER(MitarbeiterID)
);

CREATE TABLE PRODUKTKATEGORIE
(
  ProduktkategorieID INT NOT NULL,
  Kategorie VARCHAR(100) NOT NULL,
  CONSTRAINT pk_PRODUKTKATEGORIE PRIMARY KEY (ProduktkategorieID)
);

CREATE TABLE PRODUKT
(
  ProduktID INT NOT NULL,
  Produktname VARCHAR(100) NOT NULL,
  Verkaufspreis NUMERIC(9,2) NOT NULL,
  Einkaufspreis NUMERIC(9,2) NOT NULL,
  ProduktkategorieID INT NOT NULL,
  CONSTRAINT pk_PRODUKT PRIMARY KEY (ProduktID),
  CONSTRAINT fk_PRODUKT_PRODUKTKATEGORIE FOREIGN KEY (ProduktkategorieID) REFERENCES PRODUKTKATEGORIE(ProduktkategorieID)
);

CREATE TABLE BESTELLUNG_PRODUKT
(
  Menge INT NOT NULL,
  Einzelpreis NUMERIC(9,2) NOT NULL,
  BestellungID INT NOT NULL,
  ProduktID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG_PRODUKT PRIMARY KEY (BestellungID, ProduktID),
  CONSTRAINT fk_BESTELLUNG_PRODUKT_BESTELLUNG FOREIGN KEY (BestellungID) REFERENCES BESTELLUNG(BestellungID),
  CONSTRAINT fk_BESTELLUNG_PRODUKT_PRODUKT FOREIGN KEY (ProduktID) REFERENCES PRODUKT(ProduktID)
);

CREATE TABLE MITARBEITER
(
  MitarbeiterID INT NOT NULL,
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  CONSTRAINT pk_MITARBEITER PRIMARY KEY (MitarbeiterID)
);

CREATE TABLE BESTELLUNG
(
  BestellungID INT NOT NULL,
  Bestelldatum DATE NOT NULL,
  Lieferdatum DATE,
  KundenNr INT NOT NULL,
  MitarbeiterID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG PRIMARY KEY (BestellungID),
  CONSTRAINT fk_BESTELLUNG_KUNDE FOREIGN KEY (KundenNr) REFERENCES KUNDE(KundenNr),
  CONSTRAINT fk_BESTELLUNG_MITARBEITER FOREIGN KEY (MitarbeiterID) REFERENCES MITARBEITER(MitarbeiterID)
);

CREATE TABLE PRODUKTKATEGORIE
(
  ProduktkategorieID INT NOT NULL,
  Kategorie VARCHAR(100) NOT NULL,
  CONSTRAINT pk_PRODUKTKATEGORIE PRIMARY KEY (ProduktkategorieID)
);

CREATE TABLE PRODUKT
(
  ProduktID INT NOT NULL,
  Produktname VARCHAR(100) NOT NULL,
  Verkaufspreis NUMERIC(9,2) NOT NULL,
  Einkaufspreis NUMERIC(9,2) NOT NULL,
  ProduktkategorieID INT NOT NULL,
  CONSTRAINT pk_PRODUKT PRIMARY KEY (ProduktID),
  CONSTRAINT fk_PRODUKT_PRODUKTKATEGORIE FOREIGN KEY (ProduktkategorieID) REFERENCES PRODUKTKATEGORIE(ProduktkategorieID)
);

CREATE TABLE BESTELLUNG_PRODUKT
(
  Menge INT NOT NULL,
  Einzelpreis NUMERIC(9,2) NOT NULL,
  BestellungID INT NOT NULL,
  ProduktID INT NOT NULL,
  CONSTRAINT pk_BESTELLUNG_PRODUKT PRIMARY KEY (BestellungID, ProduktID),
  CONSTRAINT fk_BESTELLUNG_PRODUKT_BESTELLUNG FOREIGN KEY (BestellungID) REFERENCES BESTELLUNG(BestellungID),
  CONSTRAINT fk_BESTELLUNG_PRODUKT_PRODUKT FOREIGN KEY (ProduktID) REFERENCES PRODUKT(ProduktID)
);
````
