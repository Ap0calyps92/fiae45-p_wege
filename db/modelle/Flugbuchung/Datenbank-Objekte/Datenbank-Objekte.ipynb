# Datenbank-Objekte

Dieses Notebook verwendet SQL.



```sql
USE master;

DROP DATABASE IF EXISTS Flugbuchung;
GO

CREATE DATABASE Flugbuchung;
GO

USE Flugbuchung;

CREATE TABLE PERSON
(
  PersonID INT IDENTITY(1,1) NOT NULL,
  Vorname VARCHAR(100) NOT NULL,
  Nachname VARCHAR(100) NOT NULL,
  Geburtsdatum DATE NULL,
  CONSTRAINT pk_PERSON PRIMARY KEY (PersonID)
);

CREATE TABLE KUNDE
(
  Kundennummer CHAR(5) NOT NULL,
  Strasse VARCHAR(100) NOT NULL,
  PLZ VARCHAR(100) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  PersonID INT NOT NULL,
  CONSTRAINT pk_KUNDE PRIMARY KEY (Kundennummer),
  CONSTRAINT fk_KUNDE_PERSON FOREIGN KEY (PersonID) REFERENCES PERSON(PersonID)
);

CREATE TABLE FLUGHAFEN
(
  FlughafenCode CHAR(3) NOT NULL,
  FlughafenName VARCHAR(100) NOT NULL,
  CONSTRAINT pk_FLUGHAFEN PRIMARY KEY (FlughafenCode)
);

CREATE TABLE FLUGVERBINDUNG
(
  Flugnummer CHAR(10) NOT NULL,
  Abflugzeit TIME NOT NULL,
  Ankunftzeit TIME NOT NULL,
  Preis DECIMAL(9,2) NOT NULL,
  AbflughafenCode CHAR(3) NOT NULL,
  ZielflughafenCode CHAR(3) NOT NULL,
  CONSTRAINT pk_FLUGVERBINDUNG PRIMARY KEY (Flugnummer),
  CONSTRAINT fk_FLUGVERBINDUNG_ABFLUGHAFEN FOREIGN KEY (AbflughafenCode) REFERENCES FLUGHAFEN(FlughafenCode),
  CONSTRAINT fk_FLUGVERBINDUNG_ZIELFLUGHAFEN FOREIGN KEY (ZielflughafenCode) REFERENCES FLUGHAFEN(FlughafenCode)
);

CREATE TABLE VERKEHRSTAG
(
  Datum DATE NOT NULL,
  CONSTRAINT pk_VERKEHRSTAG PRIMARY KEY (Datum)
);

CREATE TABLE FLUG
(
  Flugnummer CHAR(10) NOT NULL,
  Datum DATE NOT NULL,
  CONSTRAINT pk_FLUG PRIMARY KEY (Flugnummer, Datum),
  CONSTRAINT fk_FLUG_FLUGVERBINDUNG FOREIGN KEY (Flugnummer) REFERENCES FLUGVERBINDUNG(Flugnummer),
  CONSTRAINT fk_FLUG_VERKEHRSTAG FOREIGN KEY (Datum) REFERENCES VERKEHRSTAG(Datum)
);

CREATE TABLE BUCHUNG
(
  BuchungID BIGINT NOT NULL,
  Buchungsdatum DATE NOT NULL,
  Kundennummer CHAR(5) NOT NULL,
  CONSTRAINT pk_BUCHUNG PRIMARY KEY (BuchungID),
  CONSTRAINT fk_BUCHUNG_KUNDE FOREIGN KEY (Kundennummer) REFERENCES KUNDE(Kundennummer)
);

CREATE TABLE BUCHUNGSPOSTEN
(
  BuchungID BIGINT NOT NULL,
  PersonID INT NOT NULL,
  Flugnummer CHAR(10) NOT NULL,
  Datum DATE NOT NULL,
  CONSTRAINT pk_BUCHUNGSPOSTEN PRIMARY KEY (BuchungID, PersonID, Flugnummer, Datum),
  CONSTRAINT fk_BUCHUNGSPOSTEN_BUCHUNG FOREIGN KEY (BuchungID) REFERENCES BUCHUNG(BuchungID),
  CONSTRAINT fk_BUCHUNGSPOSTEN_PERSON FOREIGN KEY (PersonID) REFERENCES PERSON(PersonID),
  CONSTRAINT fk_BUCHUNGSPOSTEN_FLUG FOREIGN KEY (Flugnummer, Datum) REFERENCES FLUG(Flugnummer, Datum)
);
```
