# Datenbank-Objekte

Dieses Notebook verwendet SQL.



```sql
USE master;

DROP DATABASE IF EXISTS Immobilien;
GO

CREATE DATABASE Immobilien;
GO

USE Immobilien;

CREATE TABLE EIGENTUEMER
(
  Eigentuemer_ID INT IDENTITY(1,1) NOT NULL,
  Name VARCHAR(100) NOT NULL,
  Strasse VARCHAR(100) NOT NULL,
  PLZ CHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_EIGENTUEMER PRIMARY KEY (Eigentuemer_ID)
);

CREATE TABLE IMMOBILIE
(
  Immobilie_ID INT IDENTITY(1,1) NOT NULL,
  Name VARCHAR(100) NOT NULL,
  Strasse VARCHAR(100) NOT NULL,
  PLZ CHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  Eigentuemer_ID INT NOT NULL,
  CONSTRAINT pk_IMMOBILIE PRIMARY KEY (Immobilie_ID),
  CONSTRAINT fk_IMMOBILIE_EIGENTUEMER FOREIGN KEY (Eigentuemer_ID) REFERENCES EIGENTUEMER(Eigentuemer_ID)
);

CREATE TABLE MAKLER
(
  Makler_ID INT IDENTITY(1,1) NOT NULL,
  Name VARCHAR(100) NOT NULL,
  Telefon VARCHAR(100) NOT NULL,
  CONSTRAINT pk_MAKLER PRIMARY KEY (Makler_ID)
);

CREATE TABLE KUNDE
(
  Kunde_ID INT IDENTITY(1,1) NOT NULL,
  Name VARCHAR(100) NOT NULL,
  Strasse VARCHAR(100) NOT NULL,
  PLZ CHAR(5) NOT NULL,
  Ort VARCHAR(100) NOT NULL,
  CONSTRAINT pk_KUNDE PRIMARY KEY (Kunde_ID)
);

CREATE TABLE BESICHTIGUNG
(
  Besichtigung_ID INT IDENTITY(1,1) NOT NULL,
  Besuchstermin DATE NOT NULL,
  Immobilie_ID INT NOT NULL,
  Makler_ID INT NOT NULL,
  Kunde_ID INT NOT NULL,
  CONSTRAINT pk_BESICHTIGUNG PRIMARY KEY (Besichtigung_ID),
  CONSTRAINT fk_BESICHTIGUNG_IMMOBILIE FOREIGN KEY (Immobilie_ID) REFERENCES IMMOBILIE(Immobilie_ID),
  CONSTRAINT fk_BESICHTIGUNG_MAKLER FOREIGN KEY (Makler_ID) REFERENCES MAKLER(Makler_ID),
  CONSTRAINT fk_BESICHTIGUNG_KUNDE FOREIGN KEY (Kunde_ID) REFERENCES KUNDE(Kunde_ID)
);
```
