```sql
SELECT
  i.Name AS Immobile,
  CONCAT(i.Strasse, ', ', i.PLZ, ' ', i.Ort) AS [Adresse der Immobilie],
  e.Name AS Eigentümer,
  CONCAT(e.Strasse, ', ', e.PLZ, ' ', e.Ort) AS [Adresse des Eigentümers],
  m.Name AS Makler,
  m.Telefon,
  k.Name AS Kunde,
  CONCAT(k.Strasse, ', ', k.PLZ, ' ', k.Ort) AS [Adresse des Kunden],
  FORMAT(b.Besuchstermin, 'dd.MM.yyyy') AS Besuchstermin
FROM BESICHTIGUNG b
JOIN IMMOBILIE i ON b.Immobilie_ID = i.Immobilie_ID
JOIN EIGENTUEMER e ON i.Eigentuemer_ID = e.Eigentuemer_ID
JOIN MAKLER m ON b.Makler_ID = m.Makler_ID
JOIN KUNDE k ON b.Kunde_ID = k.Kunde_ID
ORDER BY b.Besichtigung_ID;
```
