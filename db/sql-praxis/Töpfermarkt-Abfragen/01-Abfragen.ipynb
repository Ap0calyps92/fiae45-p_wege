## Zeige die Bundesländer, die mindestens 20 Märkte haben, mit der Anzahl ihrer Märkte.

Sortiere nach Anzahl absteigend.

![tm0040](.\Screenshots\tm-0040.png)

```sql
--
SELECT
  Bundesland,
  COUNT(MarktID) AS [Anzahl Märkte]
FROM Markt
GROUP BY Bundesland
HAVING COUNT(MarktID) >= 20
ORDER BY [Anzahl Märkte] DESC
```

## Wie viele Rote Krüge gibt es?

![tm0050](Screenshots/tm-0050.png)

```sql
-- Lösung mit JOINs
SELECT COUNT(*) AS [Anzahl rote Krüge]
FROM Produktart AS pa
JOIN Produkt AS p ON pa.ProduktartId = p.Produktart
JOIN Farbe AS f ON p.Farbe = f.FarbeID
WHERE pa.Bezeichnung = 'Krug' AND f.Bezeichnung = 'Rot'
```

## Zeige eine Liste der roten Gebrauchs-Krüge nach Preis aufsteigend sortiert an.

![tm0060](Screenshots/tm-0060.png)

```sql
--
SELECT
    p.ProduktCode,
    p.Preisempfehlung
FROM Produkt AS p
JOIN Produktart AS pa ON p.Produktart = pa.ProduktartId
JOIN Farbe AS f ON p.Farbe = f.FarbeId
JOIN Funktionstyp AS ft ON p.Funktionstyp = ft.FunktionstypId
WHERE pa.Bezeichnung = 'Krug'
  AND f.Bezeichnung = 'Rot'
  AND ft.Bezeichnung = 'Gebrauch'
ORDER BY p.Preisempfehlung ASC;
```

## Zeige den kleinsten, größten und mittleren empfohlenen Preis jeder Produktart an.

![tm0070](.\Screenshots\tm-0070.png)

```sql
--
SELECT
  pa.Bezeichnung AS [Produktart],
  MIN(p.Preisempfehlung) AS [kleinster Preis],
  MAX(p.Preisempfehlung) AS [größter Preis],
  AVG(p.Preisempfehlung) AS [mittlerer Preis]
FROM Produktart AS pa
JOIN Produkt AS p ON pa.ProduktartId = p.Produktart
GROUP BY pa.Bezeichnung
```

## Zeige die Bundesländer nach der Anzahl ihrer Markttermine in 2019 absteigend sortiert an.

Zeige auch Bundesländer ohne Termin an.

(21 Zeilen)

![tm0080](.\Screenshots\tm-0080.png)

```sql
--
SELECT
  m.Bundesland,
  COUNT(t.terminId) AS [Anzahl Termine 2019]
FROM Markt AS m
LEFT JOIN Termin AS t ON m.MarktId = t.MarktId AND YEAR(t.Von) = 2019
GROUP BY m.Bundesland
ORDER BY COUNT(t.terminId) DESC
```
