## 4. Zeige die Bundesländer, die mindestens 20 Märkte haben, mit der Anzahl ihrer Märkte.

Sortiere nach Anzahl absteigend.

![tm0040](.\Screenshots\tm-0040.png)

```sql
--
SELECT
  Bundesland,
  COUNT(MarktID) AS [Anzahl Märkte]
FROM Markt
GROUP BY Bundesland
HAVING COUNT(MarktID) >= 20
ORDER BY [Anzahl Märkte] DESC
```

## 6. Zeige eine Liste der roten Gebrauchs-Krüge nach Preis aufsteigend sortiert an.

![tm0060](Screenshots/tm-0060.png)

```sql
--
SELECT
    p.ProduktCode,
    p.Preisempfehlung
FROM Produkt AS p
JOIN Produktart AS pa ON p.Produktart = pa.ProduktartId
JOIN Farbe AS f ON p.Farbe = f.FarbeId
JOIN Funktionstyp AS ft ON p.Funktionstyp = ft.FunktionstypId
WHERE pa.Bezeichnung = 'Krug'
  AND f.Bezeichnung = 'Rot'
  AND ft.Bezeichnung = 'Gebrauch'
ORDER BY p.Preisempfehlung ASC;
```

## 7. Zeige den kleinsten, größten und mittleren empfohlenen Preis jeder Produktart an.

![tm0070](.\Screenshots\tm-0070.png)

```sql
--
SELECT
  pa.Bezeichnung AS [Produktart],
  MIN(p.Preisempfehlung) AS [kleinster Preis],
  MAX(p.Preisempfehlung) AS [größter Preis],
  AVG(p.Preisempfehlung) AS [mittlerer Preis]
FROM Produktart AS pa
JOIN Produkt AS p ON pa.ProduktartId = p.Produktart
GROUP BY pa.Bezeichnung
```

## 8. Zeige die Bundesländer nach der Anzahl ihrer Markttermine in 2019 absteigend sortiert an.

Zeige auch Bundesländer ohne Termin an.

(21 Zeilen)

![tm0080](.\Screenshots\tm-0080.png)

```sql
--
SELECT
  m.Bundesland,
  COUNT(t.terminId) AS [Anzahl Termine 2019]
FROM Markt AS m
LEFT JOIN Termin AS t ON m.MarktId = t.MarktId AND YEAR(t.Von) = 2019
GROUP BY m.Bundesland
ORDER BY COUNT(t.terminId) DESC
```

## 9. Finde Produkte, die im März 2019 nicht auf Märkten angeboten wurden.

![tm0090](.\Screenshots\tm-0090.png)

```sql
--SELECT *
SELECT p.ProduktCode
FROM Produkt AS p
WHERE NOT EXISTS (
    SELECT 1
    FROM Produktangebot AS pa
    JOIN Termin AS t ON pa.TerminId = t.TerminId
    WHERE pa.ProduktId = p.ProduktId
      AND YEAR(t.von) = 2019
      AND MONTH(t.von) = 3
)
```

## 10. Ranke die Farben nach Anzahl ihrer zugeordneten Produkte.

![tm0091](Screenshots/tm-0091.png)

```sql
--
SELECT
  DENSE_RANK() OVER (ORDER BY COUNT(p.ProduktId) DESC) AS [Rang],
  f.Bezeichnung,
  COUNT(p.ProduktId) AS [Anzahl Produkte]
FROM Farbe AS f
JOIN Produkt AS p ON f.FarbeId = p.Farbe
GROUP BY f.Bezeichnung
```

## 11. Zeige die Anzahl der Produkte nach Produktarten und ihren Prozentanteil an der Gesamtzahl Produkte.

![tm0100](.\Screenshots\tm-0100.png)

```sql
--
SELECT
  pa.Bezeichnung AS [Produktart],
  COUNT(pa.ProduktartId) AS [Anzahl Produkte],
  CAST(ROUND(CAST(COUNT(pa.ProduktartId) AS DECIMAL(10,4)) / SUM(COUNT(pa.ProduktartId)) OVER () * 100, 1) AS DECIMAL(10,1)) AS [Prozent Anteil an Produkten]
FROM Produkt AS p
JOIN Produktart AS pa ON p.Produktart = pa.ProduktartId
GROUP BY pa.Bezeichnung
ORDER BY [Anzahl Produkte] DESC
```

## 12. Zeige die Top 5 Bundesländer nach der Anzahl ihrer Markttermine in 2019 und den Anteil dieser Märkte an der Gesamtzahl der Termine 2019.

![tm0101](.\Screenshots\tm-0101.png)

```sql
--
SELECT
  TOP (5)
  m.Bundesland,
  COUNT(t.TerminId) AS [Anzahl Termine 2019],
  CAST(ROUND(CAST(COUNT(t.TerminId) AS DECIMAL(10,4)) / SUM(COUNT(t.TerminId)) OVER () * 100, 1) AS DECIMAL(10,1)) AS [Prozent Anteil]
FROM Markt AS m
JOIN Termin AS t ON m.MarktId = t.MarktId AND YEAR(t.Von) = 2019
GROUP BY m.Bundesland
ORDER BY [Anzahl Termine 2019] DESC
```

## 13. Ranke die Bundesländer nach Anzahl ihrer Märkte 2019

![tm0110](Screenshots/tm-0110.png)

```sql
--
SELECT
  DENSE_RANK() OVER (ORDER BY COUNT(t.TerminId) DESC) AS [Rang],
  m.Bundesland,
  COUNT(t.TerminId) AS [Anzahl Termine 2019]
FROM Markt AS m
JOIN Termin AS t ON m.MarktId = t.MarktId AND YEAR(t.Von) = 2019
GROUP BY m.Bundesland
```

## 14. Zeige Markttermine und Angebotspreise des Produkts 'KRBLG101' und die Differenz der Preise von der Preisempfehlung an.
Zeige die Abweichung zusätzlich als Bemerkung "Teurer" bzw. "Günstiger".
Sortiere nach Markttermin.
(161 Zeilen)

![tm0120](Screenshots/tm-0120.png)

```sql
--
SELECT
  CONVERT(VARCHAR, t.Von, 104) AS [Markttermin],
  p.Preisempfehlung,
  pa.Angebotspreis,
  (pa.Angebotspreis - p.Preisempfehlung) AS [Differenz],
  CASE
    WHEN (pa.Angebotspreis - p.Preisempfehlung) > 0 THEN 'Teurer'
    ELSE 'Günstiger'
  END AS [Bemerkung]
FROM Produktangebot AS pa
JOIN Termin AS t ON pa.TerminId = t.TerminId
JOIN Produkt AS p ON pa.ProduktId = p.ProduktId WHERE p.ProduktCode = 'KRBLG101'
ORDER BY t.Von
```

## 15. Finde die Markttermine, an denen das Produkt 'KRBLG101' mindestens 40 Euro billiger ist als die Preisempfehlung.

Gib die Spalten wie abgebildet aus.

![tm0130](.\Screenshots\tm-0130.png)

```sql
--
SELECT
  m.Marktbezeichnung AS [Marktbezeichnung],
  m.Ort,
  CONVERT(VARCHAR, t.Von, 104) AS [Marktbeginn],
  CONVERT(VARCHAR, t.Bis, 104) AS [Marktende],
  (p.Preisempfehlung - pa.Angebotspreis) AS [Ersparnis]
FROM Produktangebot AS pa
JOIN Termin AS t ON pa.TerminId = t.TerminId
JOIN Produkt AS p ON pa.ProduktId = p.ProduktId
JOIN Markt AS m ON t.MarktId = m.MarktId
WHERE p.ProduktCode = 'KRBLG101'
  AND (p.Preisempfehlung - pa.Angebotspreis) >= 40
ORDER BY [Ersparnis] DESC
```

## 16. Pivotiere die Anzahl der Produkte nach Produktart in Zeilen und Farbe in Spalten.

![tm0140](Screenshots/tm-0140.png)

```sql
--
SELECT
  [Produktart],
  ISNULL([Rot], 0) AS [Rot],
  ISNULL([Grün], 0) AS [Grün],
  ISNULL([Blau], 0) AS [Blau],
  ISNULL([Gelb], 0) AS [Gelb],
  ISNULL([Schwarz], 0) AS [Schwarz]
FROM (
    SELECT
      pa.Bezeichnung AS [Produktart],
      f.Bezeichnung AS [Farbe],
      p.ProduktId
    FROM Produkt AS p
    JOIN Produktart AS pa ON p.Produktart = pa.ProduktartId
    JOIN Farbe AS f ON p.Farbe = f.FarbeId
) AS Quelle
PIVOT (
    COUNT([ProduktId])
    FOR [Farbe] IN ([Rot], [Grün], [Blau], [Gelb], [Schwarz])
) AS PivotTabelle
ORDER BY [Produktart]
```
