## Zeige die Städte in Canada, wo mehr als 3 Kunden wohnen, und daneben die Anzahl der Kunden. Sortiere absteigend nach Anzahl Kunden

8 Zeilen

![kb002](Screenshots/kb002.png)

```sql
SELECT
  DISTINCT(Stadt),
  COUNT(KundenNr) AS [Anzahl Kunden]
FROM CR.Kunde
WHERE Land = 'Canada'
GROUP BY Stadt
HAVING COUNT(KundenNr) >= 3
ORDER BY [Anzahl Kunden] DESC
```

## Welche anderen Kunden wohnen in der selben Stadt wie 'Futuristic Bikes'?

Finde eine Lösung ohne Unterabfrage und eine Lösung mit Unterabfrage.

![kb003](Screenshots/kb003.png)

```sql
SELECT
  b.Firma
FROM CR.Kunde AS a
JOIN CR.Kunde AS b ON a.Stadt = b.Stadt
WHERE a.Firma = 'Futuristic Bikes' AND b.Firma <> 'Futuristic Bikes'
```

```sql
SELECT
  Firma
FROM CR.Kunde
WHERE Stadt = (SELECT Stadt FROM CR.Kunde WHERE Firma = 'Futuristic Bikes') AND Firma <> 'Futuristic Bikes'
```

## Zeige die Top 10 Kunden nach Umsatz im Jahr 2018 an. Zeige KundenNr, Firma, Anzahl Bestellungen und Umsatz im Jahr 2018 (als Currency formatiert), sortiert nach Umsatz absteigend.

Hinweis: Das Ergebnis der FORMAT-Funktion ist eine Zeichenkette, nach der nicht sinnvoll sortiert werden kann.

![kb004](Screenshots/kb004.png)

```sql
SELECT TOP 10
  k.KundenNr,
  k.Firma,
  COUNT(DISTINCT b.BestellungID) AS [Anzahl Bestellungen],
  FORMAT(SUM(bd.Bestellmenge * bd.Einzelpreis), 'C', 'de-DE') AS [Umsatz 2018]
FROM CR.Kunde AS k
JOIN Verkauf.Bestellung AS b ON k.KundenNr = b.KundenNr AND YEAR(b.Bestelldatum) = 2018
JOIN Verkauf.BestellDetail AS bd ON b.BestellungID = bd.BestellungID
GROUP BY k.KundenNr, k.Firma
ORDER BY SUM(bd.Bestellmenge * bd.Einzelpreis) DESC
```

## Was waren die kürzeste, die längste und die mittlere Bearbeitungsdauer in Tagen von Bestellung bis Versand bei allen im Jahr 2018 gelieferten Bestellungen?

Zeige den Mittelwert mit einer Nachkommastelle an.

![kb006](Screenshots/kb006.png)

```sql
SELECT
  CAST(MIN(DATEDIFF(DAY, Bestelldatum, VersandDatum)) AS decimal(10,1)) AS [kürzeste],
  CAST(MAX(DATEDIFF(DAY, Bestelldatum, VersandDatum)) AS decimal(10,1)) AS [längste],
  CAST(ROUND(AVG(CAST(DATEDIFF(DAY, Bestelldatum, VersandDatum) AS FLOAT)), 1) AS decimal(10,1)) AS [mittlere Bearbeitungsdauer]
FROM Verkauf.Bestellung
WHERE VersandDatum IS NOT NULL AND YEAR(VersandDatum) = 2018
```

## Bewerten Sie den Erfüllungszeitraum der in 2018 versandten Bestellungen wie folgt:

- weniger als 3 Tage: gut
- bis 5 Tage: ok
- darüber: schlecht

![kb007](Screenshots/kb007.png)

```sql
SELECT
  BestellungID,
  FORMAT(Bestelldatum, 'yyyy-MM-dd') AS Bestelldatum,
  FORMAT(VersandDatum, 'yyyy-MM-dd') AS VersandDatum,
  DATEDIFF(DAY, Bestelldatum, VersandDatum) AS [Tage bis Versand],
  CASE
    WHEN DATEDIFF(DAY, Bestelldatum, VersandDatum) < 3 THEN 'gut'
    WHEN DATEDIFF(DAY, Bestelldatum, VersandDatum) <= 5 THEN 'ok'
    ELSE 'Schlecht'
  END AS [Service Bewertung]
FROM Verkauf.Bestellung
WHERE YEAR(VersandDatum) = 2018 AND Status = 1
GROUP BY BestellungID, Bestelldatum, VersandDatum
```

## Berechne die Umsatzanteile der Länder der Kunden für das Jahr 2018. Ranke die Länder nach ihrem Umsatz 2018.

![kb009](Screenshots/kb009.png)

```sql
SELECT
  k.Land,
  FORMAT(CAST(SUM(bd.Bestellmenge * bd.Einzelpreis * (1 - bd.Rabatt)) AS DECIMAL(18,2)),'C', 'de-DE') AS [Umsatz 2018],
  FORMAT(CAST(SUM(bd.Bestellmenge * bd.Einzelpreis * (1 - bd.Rabatt)) * 100.0 / SUM(SUM(bd.Bestellmenge * bd.Einzelpreis * (1 - bd.Rabatt))) OVER () AS DECIMAL(10,2)), 'N2', 'de-DE') AS [Anteil %]
FROM CR.Kunde AS k
JOIN Verkauf.Bestellung AS b ON k.KundenNr = b.KundenNr AND YEAR(b.Bestelldatum) = 2018
JOIN Verkauf.BestellDetail AS bd ON b.BestellungID = bd.BestellungID
GROUP BY k.Land
```
