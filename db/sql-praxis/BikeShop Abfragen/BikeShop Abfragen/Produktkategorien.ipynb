# Auswertung nach Produktkategorien

Es gibt 42 Produktkategorien in der Datenbank. Vier Kategorien sind Oberkategorien ohne übergeordnete Kategorie. Eine weitere Kategorie ist zugleich Oberkategorie und Unterkategorie.

Produkte sind immer Unterkategorien zugeordnet, aber nicht jeder Unterkategorie sind Produkte zugeordnet.  

Es gibt 295 Produkte in der Datenbank. Achten Sie in Ihren Abfragen darauf, dass keine Produkte unberücksichtigt bleiben und keine Produkte doppelt gezählt werden.

Die Oberkategorie Vests ist selbst Unterkategorie von Clothing. Bei Abfrage von Umsätzen könnten die Umsätze sowohl Vests, als auch Clothing zugerechnet werden!
