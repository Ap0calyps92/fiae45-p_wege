# Auswertung nach Produktkategorien

Es gibt 42 Produktkategorien in der Datenbank. Vier Kategorien sind Oberkategorien ohne übergeordnete Kategorie. Eine weitere Kategorie ist zugleich Oberkategorie und Unterkategorie.

Produkte sind immer Unterkategorien zugeordnet, aber nicht jeder Unterkategorie sind Produkte zugeordnet.  

Es gibt 295 Produkte in der Datenbank. Achten Sie in Ihren Abfragen darauf, dass keine Produkte unberücksichtigt bleiben und keine Produkte doppelt gezählt werden.

Die Oberkategorie Vests ist selbst Unterkategorie von Clothing. Bei Abfrage von Umsätzen könnten die Umsätze sowohl Vests, als auch Clothing zugerechnet werden!

## Zähle alle Produktkategorien

![pk001](Screenshots/pk001.png)

```sql
SELECT
  COUNT(pk.ProduktKategorieID) AS [Anzahl Produktkategorien]
FROM Dim.ProduktKategorie AS pk
```

## Beweise, dass alle Produkte Unterkategorien zugeordnet sind

```sql
-- Keine Produkte sind Oberkategorien zugeordnet
SELECT
    p.ProduktID,
    p.Bezeichnung,
    pk.ProduktKategorieID,
    pk.Bezeichnung AS Kategorie
FROM Stamm.Produkt AS p
JOIN Dim.ProduktKategorie AS pk ON p.ProduktKategorieID = pk.ProduktKategorieID
WHERE pk.OberKategorieID IS NULL;
```

```sql
SELECT
    COUNT(*) AS [Produkte gesamt],
    SUM(CASE WHEN pk.OberKategorieID IS NOT NULL THEN 1 ELSE 0 END) AS [Produkte mit Unterkategorie]
FROM Stamm.Produkt AS p
JOIN Dim.ProduktKategorie AS pk ON p.ProduktKategorieID = pk.ProduktKategorieID;
```

## Zeige Oberkategorien und Unterkategorien, aufsteigend sortiert nach Oberkategorie und Unterkategorie

38 Zeilen

![pk002](Screenshots/pk002.png)

```sql
SELECT
    ober.Bezeichnung AS Oberkategorie,
    unter.Bezeichnung AS Unterkategorie
FROM Dim.ProduktKategorie AS unter
JOIN Dim.ProduktKategorie AS ober ON unter.OberKategorieID = ober.ProduktKategorieID
ORDER BY Oberkategorie, Unterkategorie;
```

## Zeige Oberkategorien, Unterkategorien, sowie IDs und Bezeichnungen zugeordneter Produkte, aufsteigend sortiert nach Oberkategorie, Unterkategorie und Produktbezeichnung

296 Zeilen, dabei eine Kategorie ohne Produkt

![pk003](Screenshots/pk003.png)

```sql
SELECT
    ober.Bezeichnung AS Oberkategorie,
    unter.Bezeichnung AS Unterkategorie,
    p.ProduktID,
    p.Bezeichnung AS Produkt
FROM Dim.ProduktKategorie AS unter
JOIN Dim.ProduktKategorie AS ober ON unter.OberKategorieID = ober.ProduktKategorieID
LEFT JOIN Stamm.Produkt AS p ON unter.ProduktKategorieID = p.ProduktKategorieID
ORDER BY Oberkategorie, Unterkategorie, p.ProduktID, p.Bezeichnung
;
```
