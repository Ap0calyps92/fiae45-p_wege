## Zeige IDs und Bezeichnungen **aller** Produkttypen und die Anzahl der den Typen zugeordneten Produkte. Sortiere absteigend nach der Anzahl. Auch Produkttypen ohne Zuordnung sollen angezeigt werden. (128 Zeilen)

Erstelle eine Lösung mit JOIN und eine mit korrelierter Unterabfrage.

![pt003](Screenshots/pt003.png)

```sql
-- LEFT JOIN Lösung
SELECT
  pt.*,
  COUNT(p.ProduktTypID) AS [Anzahl Produkte des Typs]
FROM Dim.ProduktTyp AS pt
LEFT JOIN Stamm.Produkt AS p ON pt.ProduktTypID = p.ProduktTypID
GROUP BY pt.ProduktTypID, p.ProduktTypID, pt.Bezeichnung
ORDER BY [Anzahl Produkte des Typs] DESC
```

## Zeige Produkttypen, Produkte und die Anzahl der verkauften Produkte im Jahr 2018, wo die Produkttypbezeichnung 'glove' enthält

Erstelle eine Lösung ohne Unterabfrage und eine Lösung mit Unterabfrage.

![pt005](Screenshots/pt005.png)

```sql
-- JOIN Lösung
SELECT
  pt.ProduktTypID AS [TypID],
  pt.Bezeichnung AS [Produkttyp],
  p.ProduktID,
  p.Bezeichnung,
  SUM(bd.Bestellmenge) AS [Stück verkauft 2018]
FROM Dim.ProduktTyp AS pt
LEFT JOIN Stamm.Produkt AS p ON pt.ProduktTypID = p.ProduktTypID
LEFT JOIN Verkauf.BestellDetail AS bd ON p.ProduktID = bd.ProduktID
LEFT JOIN Verkauf.Bestellung AS b ON bd.BestellungID = b.BestellungID
WHERE pt.Bezeichnung LIKE '%glove%'
  AND (b.Bestelldatum IS NULL OR YEAR(b.Bestelldatum) = 2018)
GROUP BY pt.ProduktTypID, pt.Bezeichnung, p.ProduktID, p.Bezeichnung
ORDER BY [Stück verkauft 2018] DESC
```

## Ranke die Produkttypen nach der Anzahl der zugeordneten Produkte

(128 Zeilen)

![pt006](.\Screenshots\pt006.png)

```sql
SELECT
  RANK() OVER (ORDER BY COUNT(p.ProduktTypID) DESC) AS Rang,
  pt.ProduktTypID,
  pt.Bezeichnung,
  COUNT(p.ProduktTypID) AS [Anzahl Produkte des Typs]
FROM Dim.ProduktTyp AS pt
LEFT JOIN Stamm.Produkt AS p ON p.ProduktTypID = pt.ProduktTypId
GROUP BY pt.ProduktTypID, pt.Bezeichnung
```

## Zeige die Produkttypen und für jeden Produkttyp die beiden meist-verkauften Produkte in 2017 

(Verkaufszahlen laut Menge in Bestelldetail)

![pt007](.\Screenshots\pt007.png)

```sql
SELECT
    ProduktTypID,
    Produkttyp,
    ProduktID,
    Produkt,
    StueckVerkauft AS [Stück verkauft in 2017]
FROM (
    SELECT
        pt.ProduktTypID,
        pt.Bezeichnung AS Produkttyp,
        p.ProduktID,
        p.Bezeichnung AS Produkt,
        SUM(bd.Bestellmenge) AS StueckVerkauft,
        ROW_NUMBER() OVER (
            PARTITION BY pt.ProduktTypID
            ORDER BY SUM(bd.Bestellmenge) DESC
        ) AS Rang
    FROM Dim.ProduktTyp AS pt
    LEFT JOIN Stamm.Produkt AS p ON pt.ProduktTypID = p.ProduktTypID
    LEFT JOIN Verkauf.BestellDetail AS bd ON p.ProduktID = bd.ProduktID
    LEFT JOIN Verkauf.Bestellung AS b ON bd.BestellungID = b.BestellungID
    WHERE b.Bestelldatum IS NULL OR YEAR(b.Bestelldatum) = 2017
    GROUP BY
        pt.ProduktTypID,
        pt.Bezeichnung,
        p.ProduktID,
        p.Bezeichnung
) AS Verkaufszahlen
WHERE Rang <= 2
ORDER BY ProduktTypID, Rang;
```
