# BikeShop Datenbank aus Backup-Datei Wiederherstellen

Nachfolgend finden Sie einen Codeblock zuw Wiederherstellung der Datenbank BikeShop auf Microsoft SQL Server 2022 in der Standard oder Developer Version.  
Für andere Versionen ist der Installationspfad im Codeblock anzupassen. 

Legen Sie die Bak-Datei auf das D-Laufwerk bzw. passen Sie den Dateipfad im Code an

```sql
-- Wiederherstellung auf SQL-Server 2022
USE [master]
DROP DATABASE IF EXISTS BikeShop;
GO
RESTORE DATABASE [BikeShop]
FROM  DISK = N'D:\BikeShop.bak'
WITH  FILE = 1,
MOVE N'BikeShop'
TO N'C:\Program Files\Microsoft SQL Server\MSSQL16.MSSQLSERVER\MSSQL\DATA\BikeShop.mdf',
MOVE N'BikeShop_log'
TO N'C:\Program Files\Microsoft SQL Server\MSSQL16.MSSQLSERVER\MSSQL\DATA\BikeShop.ldf',
NOUNLOAD,  STATS = 5;
GO
ALTER AUTHORIZATION ON DATABASE::BikeShop TO [SA];
```
