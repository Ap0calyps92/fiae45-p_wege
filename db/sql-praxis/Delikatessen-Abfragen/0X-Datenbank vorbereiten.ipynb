# Wiederherstellung der Datenbank Delikatessen

Nachfolgend finden Sie einen Codeblock zuw Wiederherstellung der Datenbank **Delikatessen** auf Microsoft SQL Server 2022 in der Standard oder Developer Version.  
Für andere Versionen ist der Installationspfad im Codeblock anzupassen. 

Legen Sie die Bak-Datei auf das D-Laufwerk bzw. passen Sie den Dateipfad im Code an

```sql
-- Wiederherstellung auf SQL-Server 2022
USE [master]
DROP DATABASE IF EXISTS Delikatessen;
GO
RESTORE DATABASE [Delikatessen]
FROM  DISK = N'D:\Delikatessen.bak'
WITH  FILE = 1,
MOVE N'Delikatessen_utf'
TO N'C:\Program Files\Microsoft SQL Server\MSSQL16.MSSQLSERVER\MSSQL\DATA\Delikatessen.mdf',
MOVE N'Delikatessen_utf_log'
TO N'C:\Program Files\Microsoft SQL Server\MSSQL16.MSSQLSERVER\MSSQL\DATA\Delikatessen.ldf',
NOUNLOAD,  STATS = 5;
GO
ALTER AUTHORIZATION ON DATABASE::Delikatessen TO [SA];
```
