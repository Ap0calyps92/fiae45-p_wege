# Delikatessen Zusatzaufgaben zur Prozentrechnung


## Zeige die Namen der Produktkategorien, den Umsatz der Kategorien und den Anteil der Umsätze in Prozent

**Hinweis:** Die erste Zeile zeigt Getränke, Umsatz 286943.95, % Anteil 21.13

```sql
SELECT
    k.bezeichnung AS [Kategorie],

    (SELECT SUM(bd.verkaufspreis * bd.anzahl)
     FROM bestelldetail AS bd
     WHERE bd.artikel_id IN (
         SELECT a.artikel_id
         FROM artikel AS a
         WHERE a.kategorie_id = k.kategorie_id
     )) AS [Umsatz],

    (SELECT SUM(bd.verkaufspreis * bd.anzahl)
     FROM bestelldetail AS bd
     WHERE bd.artikel_id IN (
         SELECT a.artikel_id
         FROM artikel AS a
         WHERE a.kategorie_id = k.kategorie_id
     )) * 100.0 /
    (SELECT SUM(bd2.verkaufspreis * bd2.anzahl)
     FROM bestelldetail AS bd2) AS [% Anteil]
FROM kategorie AS k
ORDER BY [Umsatz] DESC;
```

```sql
SELECT
    k.bezeichnung AS Kategorie,
    SUM(d.verkaufspreis * d.anzahl) AS Umsatz,
    CAST(SUM(d.verkaufspreis * d.anzahl) * 100.0 / (SELECT SUM(bd2.verkaufspreis * bd2.anzahl) FROM bestelldetail AS bd2) AS DECIMAL(5,2)) AS [% Anteil]
FROM kategorie AS k
JOIN artikel AS a ON k.kategorie_id = a.kategorie_id
JOIN bestelldetail AS d ON d.artikel_id = a.artikel_id
GROUP BY k.bezeichnung
ORDER BY Umsatz DESC;
```

## Zeige die Namen der Kunden, ihren Umsatz und den Anteil der Umsätze in Prozent. (91 Zeilen)

**Hinweis:** Die erste Zeile zeigt Kunde *Alfreds Futterkiste*, Umsatz 7253.50, % Anteil 0.53

```sql
SELECT
    k.firma AS Kunde,
    SUM(bd.verkaufspreis * bd.anzahl) AS Umsatz,
    CAST(SUM(bd.verkaufspreis * bd.anzahl) * 100.0 / (SELECT SUM(bd.verkaufspreis * bd.anzahl) FROM bestelldetail AS bd) AS DECIMAL(5,2)) AS [% Anteil]
FROM Kunde AS k
LEFT JOIN bestellung AS b ON k.kunde_id = b.kunde_id
LEFT JOIN bestelldetail AS bd ON bd.bestellung_id = b.bestellung_id
GROUP BY k.firma
ORDER BY k.firma;
```

## Zeige Mitarbeiter, die Anzahl bearbeiteter Bestellungen und den entsprechenden Prozentanteil

**Hinweis:** Die erste Zeile zeigt Mitarbeiter *Fuller, Andrew* mit 96 Bestellungen und % Anteil 11.50

```sql
SELECT
    p.nachname + ', ' + p.vorname AS Mitarbeiter,
    COUNT(b.personal_id) AS Anzahl,
    CAST(COUNT(b.personal_id) * 100.0 / (SELECT COUNT(*) FROM bestellung) AS DECIMAL(5,2)) AS [% Anteil]
FROM personal AS p
JOIN bestellung AS b ON p.personal_id = b.personal_id
GROUP BY p.nachname, p.vorname
ORDER BY p.vorname
```

## Zeige die Namen der Produktkategorien, den Umsatz der Kategorien im Jahr 2013 und den Anteil der Umsätze in Prozent

Sortiere nach Produktkategorie.

**Hinweis:** Die erste Zeile zeigt Produktkategorie *Fleischprodukte* mit Umsatz 2013: 87621.03 und % Anteil 13.31

```sql
SELECT
    k.bezeichnung AS Kategorie,
    SUM(bd.verkaufspreis * bd.anzahl) AS [Umsatz 2013],
    CAST(SUM(bd.verkaufspreis * bd.anzahl) * 100.0 / (SELECT SUM(bd2.verkaufspreis * bd2.anzahl) FROM bestelldetail AS bd2 JOIN bestellung AS b2 ON bd2.bestellung_id = b2.bestellung_id WHERE YEAR(b2.bestelldatum) = 2013) AS DECIMAL(5,2)) AS [% Anteil]
FROM kategorie AS k
JOIN artikel AS a ON k.kategorie_id = a.kategorie_id
JOIN bestelldetail AS bd ON bd.artikel_id = a.artikel_id
JOIN bestellung AS b ON bd.bestellung_id = b.bestellung_id
WHERE YEAR(b.bestelldatum) = 2013
GROUP BY k.bezeichnung
ORDER BY k.bezeichnung
```

## Zeige die Namen der Kunden, ihren Umsatz im Jahr 2013 und den Anteil der Umsätze in Prozent

87 Zeilen: nicht alle Kunden hatten Umsatz!

**Hinweis:** Die erste Zeile zeigt Kunde *Alfreds Futterkiste*, Umsatz in 2013: 4873.35, % Anteil 0.74

```sql
SELECT
    k.firma AS Kunde,
    COALESCE(SUM(bd.verkaufspreis * bd.anzahl), 0) AS [Umsatz 2013],
    CAST(ROUND(COALESCE(SUM(bd.verkaufspreis * bd.anzahl), 0) * 100.0 / (SELECT SUM(bd2.verkaufspreis * bd2.anzahl) FROM bestellung AS b2 JOIN bestelldetail AS bd2 ON b2.bestellung_id = bd2.bestellung_id WHERE YEAR(b2.bestelldatum) = 2013), 2) AS DECIMAL(5,2)) AS [Prozent Anteil]
FROM kunde AS k
JOIN bestellung AS b
    ON k.kunde_id = b.kunde_id
    AND YEAR(b.bestelldatum) = 2013
JOIN bestelldetail AS bd
    ON b.bestellung_id = bd.bestellung_id
GROUP BY k.kunde_id, k.firma
ORDER BY [Umsatz 2013] DESC;
```

## Zeige Mitarbeiter, die Anzahl in 2013 bearbeiteter Bestellungen und den entsprechenden Prozentanteil

**Hinweis:** Die erste Zeile zeigt Mitarbeiter *Fuller, Andrew* mit 41 Bestellungen in 2013 und % Anteil 10.05

```sql
SELECT
    CONCAT(p.nachname, ', ', p.vorname) AS Mitarbeiter,
    COUNT(b.bestellung_id) AS [Bestellungen 2013],
    CAST(COUNT(b.bestellung_id) * 100.0 / SUM(COUNT(b.bestellung_id)) OVER () AS DECIMAL(5,2)) AS [% Anteil]
FROM personal AS p
JOIN bestellung AS b
    ON p.personal_id = b.personal_id
    AND YEAR(b.bestelldatum) = 2013
GROUP BY p.nachname, p.vorname;
```
