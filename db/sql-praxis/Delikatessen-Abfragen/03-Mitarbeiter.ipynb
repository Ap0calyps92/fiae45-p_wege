# Abfragen nach Mitarbeitern

## 1. Zeige jede mögliche Kombination aus Mitarbeiter-Vornamen und Mitarbeiter-Nachnamen an.

Nummeriere die Ausgabe, sortiert nach Nachname und Vorname. Verkette Vor- und Nachname zu einer Spalte "Mitarbeiter Mix". **(81 Zeilen)**

![Mi-0010](.\Screenshots\Mi-0010.png)

```sql
--
SELECT
  ROW_NUMBER() OVER (ORDER BY p2.nachname, p1.vorname) AS Nr,
  CONCAT(p1.vorname, ' ', p2.nachname) AS [Mitarbeiter Mix]
FROM personal AS p1
CROSS JOIN personal AS p2
ORDER BY p2.nachname, p1.vorname
```

## 2. Zeige Mitarbeiter an, die insgesamt über 100 Bestellungen bearbeitet haben.

Sortiere nach Anzahl Bestellungen absteigend.

![Mi-0020](Screenshots/Mi-0020.png)

```sql
--
SELECT p.vorname, COUNT(b.bestellung_id) AS [Anzahl Bestellungen]
FROM personal AS p
LEFT JOIN bestellung AS b ON p.personal_id = b.personal_id
GROUP BY p.vorname
HAVING COUNT(b.bestellung_id) > 100
ORDER BY [Anzahl Bestellungen] DESC
```

## 3. Zeige die Namen aller Mitarbeiter an und die Anzahl **verschiedener** Kunden, deren Bestellungen der jeweilige Mitarbeiter in 2013 bearbeitet hat.

![Mi-0030](.\Screenshots\Mi-0030.png)

```sql
--
SELECT p.vorname, COUNT(DISTINCT b.kunde_id) AS [Verschiedene Kunden 2013]
FROM personal AS p
LEFT JOIN bestellung AS b ON p.personal_id = b.personal_id AND YEAR (b.bestelldatum) = 2013
GROUP BY p.vorname
ORDER BY [Verschiedene Kunden 2013] DESC
```

## 4. Zeige Mitarbeiter und ihre Untergebenen an.

Joine hierzu die Tabelle **Personal** mit sich selbst.

![Mi-0040](Screenshots/Mi-0040.png)

```sql
--
SELECT
  p1.vorname AS [Mitarbeiter],
  p1.position as [Mitarbeiter Position],
  p2.vorname AS [Untergebener],
  p2.position AS [Untergebene Position]
FROM personal AS p1
LEFT JOIN personal AS p2 ON p2.vorgesetzte = p1.personal_id
ORDER BY p1.vorgesetzte
```
