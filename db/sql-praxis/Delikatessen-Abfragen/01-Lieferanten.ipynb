# Abfragen nach Lieferanten

## 1. Zeige Lieferanten aus der Region **Québec** an.

![lief-0010](Screenshots/lief-0010.png)

```sql
--
SELECT firma as "Lieferant", Ort, Region, Land
FROM lieferant
WHERE Region = 'Québec';
```

## 2. Zeige die Bezeichnungen, Liefereinheiten und Preise der Artikel an, die **Bigfoot Breweries** liefert.

![lief-0020](Screenshots/lief-0020.png)

```sql
--
SELECT a.bezeichnung AS 'Artikel', a.liefereinheit AS 'Einheit', a.einzelpreis AS 'Preis'
FROM artikel a
JOIN lieferant l ON a.lieferant_id = l.lieferant_id
WHERE l.firma = 'Bigfoot Breweries';
```

## 3. Zeige alphabetisch aufsteigend sortiert die Firmennamen aller Lieferanten mit der Anzahl der Produkte an, die sie liefern. (29 Zeilen)

![lief-0030](Screenshots/lief-0030.png)

```sql
-- JOIN Lösung
SELECT l.firma AS 'Lieferant', COUNT(a.artikel_id) AS 'Anzahl Produkte'
FROM lieferant l
JOIN artikel a ON l.lieferant_id = a.lieferant_id
GROUP BY l.firma
ORDER BY l.firma ASC;
```

```sql
-- Korrelierte Unterabfrage
SELECT
    l.firma AS 'Lieferant',
    (SELECT COUNT(*)
     FROM artikel AS a
     WHERE a.lieferant_id = l.lieferant_id) AS 'Anzahl Produkte'
FROM lieferant AS l
ORDER BY l.firma ASC;
```

## 4. Ranke alle Lieferanten nach der Anzahl Produkte, die sie liefern. (29 Zeilen)

![lief-0040](Screenshots/lief-0040.png)

```sql
-- JOIN Lösung
SELECT
    l.firma AS 'Lieferant',
    COUNT(a.artikel_id) AS 'Anzahl Produkte',
    DENSE_RANK() OVER (ORDER BY COUNT(a.artikel_id) DESC) AS 'Rang'
FROM lieferant l
JOIN artikel a ON l.lieferant_id = a.lieferant_id
GROUP BY l.firma
ORDER BY Rang;
```
