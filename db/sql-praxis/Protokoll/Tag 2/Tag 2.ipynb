## Was passiert bei JOIN?

- man kann **Spalten** aus mehreren Tabellen in einer Abfrage benutzen

## Was passiert mit den Zeilen beim JOIN?

- INNER JOIN: nur Zeilen aus beiden Tabellen, die zueinander passen
- LEFT OUTER JOIN: alle Zeilen aus der ersten Tabelle und passende aus der anderen
- RIGHT OUTER JOIN: alle Zeilen aus der zweiten Tabelle und passende aus der anderen
- FULL OUTER JOIN: alle Zeilen aus beiden Tabellen

## Was passiert mit den Spalten der jeweils anderen Tabelle, wenn sie keine passenden Zeilen hat?

- Die Spalten der anderen Tabelle, die keine passenden Zeilen hat, werden auf NULL gesetzt

## Situationen für OUTER JOIN

- die Aufgabe verlangt ausdrücklich, alle Spalten anzuzeigen



## Besonderheiten bei Filtern von OUTER JOINs

### Bei Left JOINs Filtern in einer Spalte der RIGHT Tabelle: Filter in ON Klausel, nicht WHERE

WHERE Klausel macht LEFT JOIN kaputt

```sql
SELECT k.Firma, b.Bestelldatum
FROM Kunde AS k
LEFT JOIN Bestellung AS b ON k.KundenNR = b.KundenNr
WHERE YEAR(b.Bestelldatum) = 2019 -- macht LEFT JOIN kaputt
```

Filter in der ON Klausel für Spalte in der RIGHT Tabelle beim LEFT JOIN

```sql
SELECT k.Firma, b.Bestelldatum
FROM Kunde AS k
LEFT JOIN Bestellung AS b ON k.KundenNR = b.KundenNr AND YEAR (b.Bestelldatum) = 2019
```

### Bei RIGHT JOINs Filtern in einer Spalte der LEFT Tabelle: Filter in ON Klausel, nicht WHERE

WHERE Klausel macht RIGHT JOIN kaputt

````sql
-- INNER JOIN
SELECT k.Firma, COUNT(b.BestellungID) AS [Anzahl Bestellungen]
FROM KUNDE AS k
JOIN Bestellung AS b ON k.KundenNr = b.KundenNr
GROUP BY k.Firma;
````

````sql
-- LEFT JOIN
SELECT k.Firma, COUNT(b.BestellungID) AS [Anzahl Bestellungen]
FROM KUNDE AS k
LEFT JOIN Bestellung AS b ON k.KundenNr = b.KundenNr
GROUP BY k.Firma
ORDER BY [Anzahl Bestellungen] ASC;
````

````sql
-- RIGHT JOIN
SELECT k.Firma, COUNT(b.BestellungID) AS [Anzahl Bestellungen]
FROM KUNDE AS k
RIGHT JOIN Bestellung AS b ON k.KundenNr = b.KundenNr
GROUP BY k.Firma
ORDER BY [Anzahl Bestellungen] ASC;
````

````sql
SELECT k.Firma, b.Bestelldatum
FROM Kunde AS k
LEFT JOIN Bestellung AS b ON k.KundenNr = b.KundenNr
WHERE b.Bestelldatum IS NULL
````

````sql
SELECT k.Firma, b.Bestelldatum
FROM Kunde AS k
LEFT JOIN Bestellung AS b ON k.KundenNr = b.KundenNr AND YEAR (b.Bestelldatum) = 2019
````
