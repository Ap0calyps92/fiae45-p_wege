# Tag 3

- Abgleich zwischen Tabellen mit FULL JOIN
- Mengenoperationen mit UNION, UNION ALL, INTERSECT, EXCEPT
- Alte JOIN Syntax
- Unterabfragen in verschiedenen Klauseln

````sql
SELECT a.Vorname, a.Nachname, a.Ort
FROM Adresse AS a TABLESAMPLE SYSTEM(10 PERCENT) REPEATABLE(456)
````

````sql
-- Demo Abgleich von Tabellen mit FULL OUTER JOIN

SELECT *
INTO KundenSample1
FROM Kunde TABLESAMPLE SYSTEM (80 PERCENT) REPEATABLE (123);
````

````sql
SELECT *
INTO KundenSample2
FROM Kunde TABLESAMPLE SYSTEM (80 PERCENT) REPEATABLE (456);
````

````sql
SELECT
  s1.KundenNr AS K1,
  s1.Firma AS F1,
  s2.KundenNr AS K2,
  s2.Firma AS F2
FROM KundenSample1 AS s1
FULL JOIN KundenSample2 AS s2 ON s1.KundenNr = s2.KundenNr
WHERE s1.KundenNr IS NULL
````

## Mengenoperationen

- Zwei komplette Abfragen mit gleicher Anzahl Spalten und entsprechenden Datentypen

- UNION Vereinigungsmenge
- UNION ALL mit Doppelungen
- INTERSECT Schnittmenge
- EXCEPT Differenzmenge

````sql
SELECT KundenNr, Firma FROM Kundensample1
UNION ALL
SELECT KundenNr, Firma FROM Kundensample2
````

````sql
SELECT KundenNr, Firma FROM Kundensample1
EXCEPT
SELECT KundenNr, Firma FROM Kundensample2
````

## POV: 80er Jahre Datenbanktyp kann nur alte JOIN Syntax und muss folgendes auf MS-SQL-Server lösen

Liste **alle** Kunden und die Anzahl ihrer Bestellungen: LEFT JOIN

```sql
SELECT Kunde.KundenNR, Firma, COUNT(BestellungID) AS [Anzahl Bestellungen]
FROM Kunde, Bestellung
WHERE Kunde.KundenNR *= Bestellung.KundenNr
GROUP BY Kunde.KundenNR, Firma;

````sql
-- LEFT JOIN in aktueller Syntax
SELECT
  k.KundenNr,
  k.firma,
  COUNT(b.BestellungID) AS [Anzahl Bestellungen]
FROM Kunde AS k
LEFT JOIN Bestellung AS b ON k.KundenNr = b.KundenNr
GROUP BY k.KundenNr, k.Firma
ORDER BY [Anzahl Bestellungen] ASC;
````

````sql
SELECT
  k.KundenNr,
  k.Firma,
  (SELECT COUNT(*)
  FROM Bestellung AS b
  WHERE k.KundenNr = b.KundenNr
  ) AS [Anzahl Bestellungen]
  FROM Kunde AS k
  ORDER BY [Anzahl Bestellungen] ASC;
````
