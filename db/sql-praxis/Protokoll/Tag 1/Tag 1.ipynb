# Tag 1

Der Schwerpunkt des Moduls ist **Abfragen mit Microsoft SQL Server**.

Wichtige Teilaspekte:

- Welche Klauseln gehören zu einer Abfrage
- Projektion: Spaltenauswahl, Spalten-Aliase (Spaltenüberschrift)
- Selektion: Zeilenauswahl, Filter
- Filter in verschiedenen Klauseln: WHERE, HAVING, ON
- Unterschiedliche Arten von JOINs
- Berechnungen: Aggregationen (SUM, MAX, MIN, AVG)
- Unterabfragen: Abfragen verschachteln
- Mengenoperationen: UNION (ohne Duplikate), UNION ALL (mit Duplikaten), INTERSECT (Schnittmenge), EXCEPT (MINUS)

Zu vertiefende Themen abseits der Abfragen:

- DDL: Objekte erstellen und manipulieren
- DCL: Berechtigungen in der Datenbank steuern
