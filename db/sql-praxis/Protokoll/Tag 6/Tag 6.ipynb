## Referentielle Integrität

FOREIGN KEYs schützen die Integrität abhängiger Datensätze, so dass ein referenzierter Primary Key nicht geändert und nicht gelöscht werden kann.

Die einzig empfehlenswerte Standardeinstellung ist `ON UPDATE NO ACTION ON DELETE NO ACTION`.

Folgende Alternativen sind möglich:

- `CASCADE`
- `SET NULL`
- `SET DEFAULT`
