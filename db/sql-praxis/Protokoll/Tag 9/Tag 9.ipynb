  - `UPSERT`
- Abfrageergebnisse als XML oder JSON ausgeben

## XML Syntax

- Besteht aus **Elementen** und **Attributen**
- Auf höchster Ebene gibt es ein **Wurzelelement**
- Elemente haben typisch ein **Anfangs-** und ein **End-Tag**
- End-Tags haben einen Slash nach der öffnenden Klammer
- Elemente, die nichts umschließen, heißen **leere Elemente**
- Leere Elemente werden mit einem Slash im Anfangs-Tag geschlossen
- Elemente können im Anfangs-Tag zusätzlich Daten in Attributen haben.
- Beim Entwurf vom XML-Strukturen entscheidet der Autor, welche Daten in Elemente, und welche in Attribute geschrieben werden.
- Unverbindliche Faustregel:
  - Metadaten in Attribute
  - Harte Fakten in Elemente

### XML Syntaxregeln

XML-Dateien, die grundlegende Syntaxregel befolgen, sind **wohlgeformt** (**well-formed**). Wenn eine Datei nicht wohlgeformt ist, kann sie nicht geladen werden.

- Die XML-Datei hat nur ein Wurzelelement.
- Attributwerte stehen in Anführungszeichen.
- XML ist case-sensitiv.
- Elemente dürfen nicht über Kreuz verschachtelt werden.
- Namen von Elementen und Attributen dürfen nicht mit einer Ziffer beginnen und keine Leerzeichen enthalten.
- XML-eigene Zeichen müssen maskiert werden. Diese Maskierungs-Ausdrücke heißen **named Entities**
  - `<` als `&lt;`
  - `>` als `&gt;`
  - `&` als `&amp;`
  - `"` als `&quot;`
  - `'` als `&apos;`

### JSON: JavaScript Object Notation

- enthält **Arrays** in eckigen Klammern
- Array-Elemente werden durch Komma getrennt
- enthält **Objekte** in geschweiften Klammern
- Objekte haben **Properties** (Eigenschaften, Attribute) und **Values** (Werte)
- Attribute und Werte werden duch Doppelpunkte getrennt
- Attribut-Wert-Paare werden durch komma getrennt
- Werte können einfache Datentypen wie String, oder Number sein, aber auch Arrays oder Objekte
- Zeichenketten, Datum und Uhrzeit stehen in Anführungszeichen oder Hochkomma
- Integer und Dezimalzahlen ohne Anführungszeichen
- Dezimalzahlen ohne Einheuten und imme rmit Dezimalpunkt
